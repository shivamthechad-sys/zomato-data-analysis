**Zomato Project**

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

**Step 2 - Create the data frame**

```python
dataframe = pd.read_csv("Zomato data .csv")
print(dataframe)
```

```python
dataframe
```

**Convert the data type of column - rate**

```python
def handleRate(value):
    value = str(value).split('/')
    value = value[0];
    return float(value)

dataframe['rate']=dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
dataframe.info()
```

**Type of resturant**

```python
dataframe.head()
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("type of resturant")
              
```

# Majority of the rstaurant falls in dinning category

```python
 dataframe.head()
```

```python
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})
plt.plot(result, c="green", marker="o")
plt.xlabel("Type of restaurant" , c="red", size=20)
plt.ylabel("votes", c="red", size=20)                      
```

**Conclusion - Dinning resturants has received maximum votes**

# Average order spending of couples

```python
dataframe.head()
```

```python
plt.figure(figsize=(14,6))

sns.countplot(
    x='approx_cost(for two people)',
    data=dataframe,
    hue='online_order',
    palette='Set2'
)

plt.xticks(rotation=45)
plt.title("Cost vs Online Order")
plt.show()
```

**Conclusion - The majority of couples preferr resturants with an approximate cost of 300 ruppees.**

# Which mode receives maximum rating 

```python
plt.figure(figsize=(6,6))

sns.boxplot(
    x='online_order',
    y='rate',
    data=dataframe,
    palette='Set2'
)

plt.title("Restaurant Rating vs Online Order")
plt.xlabel("Online Order")
plt.ylabel("Rating")
plt.show()

```

**Coclusion - Offline order received lower rating in comparison to online order** 

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap="YlGnBu", fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Listed In (Type)")
plt.show()
```

**Conclusion - Dinning restaurants primarily accept offline orders, whereas cafes primarily rweceive online orders. This suggests thst clints prefer offline orders in person at resaturants, but prefer online ordering at cafes.**
